```python
from google.colab import userdata
import os

os.environ['LANGCHAIN_TRACING_V2'] = 'true'
os.environ['LANGCHAIN_API_KEY'] = userdata.get('LANGSMITH_KEY')
os.environ['LANGCHAIN_PROJECT'] = 'Lab9-Agentic-AI'
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(api_key= userdata.get('Agentic'),
               model="llama-3.1-8b-instant", temperature=0)

question = ('A shop sells pens at 12 rupees each. '
            'If I buy 7 and pay with 100, how much change? '
            'Think step by step.')
print(llm.invoke(question).content)
```
